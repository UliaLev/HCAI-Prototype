```python
import pandas as pd
import numpy as np

with open("../data/processed/recipes_final_dataset2.json", "r", encoding="utf-8") as file:
    df_recipes_final = pd.read_json(file)
print(f"Loaded {len(df_recipes_final)} recipes.")
```

```python
cols_to_drop = ["instructions_flat", "instr_len", "title_clean"]
df = df_recipes_final.drop(columns=cols_to_drop)

df['tags'].apply(lambda d: d.pop('meal', None) if isinstance(d, dict) else d)

print(df.iloc[0]) # Verify 
```

```python
df.to_json("../data/processed/double_cleaned_recipes.json", 
                orient='records', 
                indent=2, 
                force_ascii=False)

print("Saved successfully")
```
