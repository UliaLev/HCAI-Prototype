```python
import pandas as pd

dataset = "../data/raw/recipes_images.json"

dataset = "../data/processed/recipes_final_dataset.json"

with open(dataset, "r", encoding="utf-8") as file:
    recipes = pd.read_json(file)

tag_attributes = [
    "type",
    "cuisine",
    "ingredient",
    "meal",
    "special-consideration",
    "simple-cooking",
]

rows = []

for recipe in recipes:
    tags = recipe["tags"] or {}

    for attribute in tag_attributes:
        for value in tags.get(attribute, []):
            rows.append({
                "attribute": attribute,
                "value": value
            })

df = pd.DataFrame(rows)

result = (
    df.drop_duplicates()
    .sort_values(["attribute", "value"])
    .groupby("attribute")["value"]
    .apply(list)
    .reset_index()
)

pd.set_option("display.max_rows", None)
pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", None)
pd.set_option("display.width", None)

#result_long = result["value"].reset_index(drop=True)
print(result)

result.to_csv(
    "../data/processed/unique_tag_values_final.csv",
    index=False
)
```

```python
import pandas as pd

dataset = "../data/processed/recipes_final_dataset.json"

recipes = pd.read_json(dataset)

tag_attributes = [
    "type",
    "cuisine",
    "ingredient",
    "meal",
    "special-consideration",
    "simple-cooking",
]

rows = []

for _, recipe in recipes.iterrows():
    tags = recipe["tags"]

    if not isinstance(tags, dict):
        continue

    for attribute in tag_attributes:
        for value in tags.get(attribute, []):
            rows.append({
                "attribute": attribute,
                "value": value
            })

df = pd.DataFrame(rows)

result = (
    df.drop_duplicates()
      .sort_values(["attribute", "value"])
      .groupby("attribute")["value"]
      .apply(list)
      .reset_index()
)

pd.set_option("display.max_rows", None)
pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", None)
pd.set_option("display.width", None)

print(result)

result.to_csv(
    "../data/processed/unique_tag_values_final.csv",
    index=False
)
```
