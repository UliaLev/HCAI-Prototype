```python
import pandas as pd

df = pd.read_json("../data/processed/recipes_group_preprocessed.json")

grouped_values = {}

for field in ["cuisine_grouped", "ingredient_grouped", "type_grouped"]:
    grouped_values[field] = sorted(
        df["tags"]
        .apply(lambda x: x.get(field, []))
        .explode()
        .dropna()
        .unique()
        .tolist()
    )

print(f"['cuisine_grouped']: {grouped_values['cuisine_grouped']}")
print(f"['ingredient_grouped']: {grouped_values['ingredient_grouped']}")
print(f"['type_grouped']: {grouped_values['type_grouped']}")
```
